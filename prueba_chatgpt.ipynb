```python
import torch
from diffusers import StableDiffusionXLPipeline

print("⏳ Cargando SDXL...")

pipe = StableDiffusionXLPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    use_safetensors=True,
    low_cpu_mem_usage=True
)

pipe.enable_model_cpu_offload()
pipe.enable_attention_slicing()

print("✅ SDXL está listo")
```

```python
def analizar_categoria(imagen, candidatos, cantidad=3):
    resultados = modelo_clip(
        imagen,
        candidate_labels=candidatos
    )

    return resultados[:cantidad]


def analizar_objeto(imagen):

    # Objeto
    candidatos_objetos = [
        "a light bulb",
        "a bottle",
        "a cup",
        "a mug",
        "a book",
        "a pen",
        "a pencil",
        "a phone",
        "a lamp",
        "a plant",
        "a flower",
        "a vase",
        "a spoon",
        "a fork",
        "a plate",
        "a bowl",
        "a toy",
        "a clock",
        "a watch",
        "a pair of glasses",
        "a remote control",
        "a shoe",
        "a key",
        "a box",
        "a household object"
    ]

    objeto = analizar_categoria(
        imagen,
        candidatos_objetos,
        cantidad=1
    )

    # Forma
    formas = [
        "a round object",
        "a spherical object",
        "an oval object",
        "a rectangular object",
        "an elongated object",
        "a cylindrical object",
        "an irregular object"
    ]

    forma = analizar_categoria(imagen, formas)

    # Color
    colores = [
        "a white object",
        "a black object",
        "a gray object",
        "a red object",
        "a blue object",
        "a green object",
        "a yellow object",
        "a brown object",
        "a transparent object",
        "a multicolored object"
    ]

    color = analizar_categoria(imagen, colores)

    # Material
    materiales = [
        "a plastic object",
        "a glass object",
        "a metal object",
        "a ceramic object",
        "a wooden object",
        "a paper object",
        "a fabric object",
        "a stone object",
        "a shiny object",
        "a matte object"
    ]

    material = analizar_categoria(imagen, materiales)

    # Textura
    texturas = [
        "a smooth surface",
        "a rough surface",
        "a textured surface",
        "a glossy surface",
        "a translucent surface",
        "a soft surface",
        "a hard surface"
    ]

    textura = analizar_categoria(imagen, texturas)

    return {
        "objeto": objeto,
        "forma": forma,
        "color": color,
        "material": material,
        "textura": textura
    }
```

```python
perfil = analizar_objeto(imagen)

print("🔎 PERFIL VISUAL")
print("=" * 40)

for categoria, resultados in perfil.items():
    print(f"\n{categoria.upper()}")

    for resultado in resultados:
        print(
            f"  • {resultado['label']} "
            f"→ {resultado['score']:.2%}"
        )
```
